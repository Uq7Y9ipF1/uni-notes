## Цель
Изучение PostgreSQL.


```python
import psycopg2
def execute(q):
    conn = psycopg2.connect(host='localhost', database='students', user='postgres', password='pwd')
    cur = conn.cursor()
    cur.execute(q)
    r = cur.fetchall() if q.strip().upper().startswith('SELECT') else cur.rowcount
    conn.commit(); conn.close()
    return r
```

## Задания


```python
q = '''SELECT COUNT(*) FROM student;'''
print(execute(q))
```
